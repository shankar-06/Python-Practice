### File Handling
#### 1. The Golden Rule: The with Statement
* While you can open a file using file = open("data.txt") and close it using file.close(), the best practice in Python is to use a with statement (also called a context manager).

* The with statement automatically closes the file for you when the block of code finishes—even if an error occurs while the file is open.

#### 2. File Modes
When you open a file, you must tell Python what you want to do with it by passing a "mode" string.

* "r"	Read (Default)	Opens a file for reading. Errors if the file doesn't exist.
* "w"	Write	Opens for writing. Overwrites the file if it exists, or creates a new one.
* "a"	Append	Opens for writing. Adds new data to the end of the file without overwriting.
* "r+"Read & WriteOpens for both reading and writing.

#### Read Mode: 
#### The Four File Reading Methods
##### 1. file.read(size) - The "All at Once" Method
This method reads the file into a single, continuous string.

How it works: If you call .read(), it reads from the current cursor position all the way to the end of the file.

The size argument: You can pass a number to read a specific number of characters. For example, .read(10) reads exactly 10 characters and stops.

Best for: Small files where you want to search the entire text at once (like counting how many times a specific word appears).

Warning: Loading a 10 GB file with .read() will crash your program because it runs out of RAM.

##### 2. file.readline() - The "One Line at a Time" Method
This method reads a single line of text up to (and including) the hidden newline character \n, then stops.

How it works: Every time you call it, it fetches the next line and moves the cursor to the start of the line after that.

Best for: When you only care about the very first line of a file (like reading column headers in a dataset), or when manually controlling a loop.

##### 3. file.readlines() - The "List" Method
This method reads the entire file and chops it up into a Python list. Each item in the list is one line from the file.

How it works: Under the hood, it calls .readline() repeatedly until the file ends, packing the results into a list like ["Line 1\n", "Line 2\n"].

Best for: When you need to access a specific line by its index (e.g., jumping straight to the 5th line using lines[4]) or sorting the lines.

Warning: Like .read(), this loads the entire file into memory at once.

##### 4. for line in file: - The "Iterator" Method (Industry Standard)
This is not a specific function, but the built-in way Python handles file objects in loops.

How it works: It dynamically fetches one line, lets you process it, and discards it from memory before fetching the next line.

Best for: Processing massive files securely and efficiently. This is the gold standard for reading logs or datasets.

#### Why use readlines()?
You use readlines() when you need to manipulate the structure of the file's contents in memory before doing something with it.

Because readlines() converts the entire file into a Python list (e.g., ["Line 1\n", "Line 2\n"]), you unlock all of Python's powerful list features:

Direct Indexing: You want to jump straight to the 5th line (lines[4]) without looping through the first four.

Reversing: You want to read a file from bottom to top (e.g., reading the most recent logs first using lines[::-1]).

Sorting: You want to alphabetize the lines in a file by sorting the list before writing it back.

readlines() keeps the \n attached to the end of every string in the list.

#### Write Mode: 
#### 1. The Three Write Modes
When writing data, you must choose exactly how Python should treat the file if it already exists or if it is completely new.

"w" (Write): The Bulldozer.

If the file exists, it completely erases the old content before writing.

If the file does not exist, it creates a new one.

"a" (Append): The Add-On.

If the file exists, it leaves the old data safely alone and places the cursor at the very end to add new text.

If the file does not exist, it creates a new one.

"x" (Exclusive Create): The Safe Guard.

Creates a new file and writes to it.

If the file already exists, it crashes and throws a FileExistsError. This is used when you absolutely want to avoid accidentally overwriting an existing file.

#### 2. The Two Write Methods
Unlike reading (which has four methods), writing only has two built-in methods.

A. file.write(string) - Single String Insertion
Writes exactly what you pass into it.

The Catch: It does not automatically add a new line. If you write three times in a row, it squishes everything together on one line unless you manually type \n.

Python
with open("test.txt", "w") as f:
    f.write("Hello")
    f.write("World") 
    # Output inside file: HelloWorld
    
B. file.writelines(list) - List Insertion
Takes a list of strings and writes them to the file consecutively.

The Catch: Just like .write(), it does not add newlines between the list items. You must ensure your list items already have \n at the end of them.

Python
lines = ["Line 1\n", "Line 2\n", "Line 3\n"]
with open("test.txt", "w") as f:
    f.writelines(lines)

#### Why use writelines()?
You use writelines() for bulk operations.

If you already have a Python list full of strings (maybe you generated it in your code, or maybe you got it from readlines()), it is much cleaner and faster to dump the entire list into the file at once using writelines(my_list) rather than writing a for loop to write them one by one.

writelines() expects your list items to already have \n attached; otherwise, it will squish them all onto one line.

#### Append Mode: 
#### 1. Append Mode ("a") Mechanics
When you open a file in "a" mode:

If the file exists: Python places the virtual cursor at the very end of the document. Any .write() commands will start adding text there.

If the file does not exist: Python safely creates a brand new file (exactly like "w" mode does).

Python
with open("system_log.txt", "a") as file:
    file.write("New event recorded.\n")

##### Exercise 3: The Config Reverser (readlines() + writelines())
The Goal: This tests the synergy between the two methods. Create a file called boot_order.txt containing three lines:
System Check
Load Drivers
Launch UI

Write a script that:

Opens boot_order.txt in "r" mode and uses .readlines() to pull the lines into a list.

Reverses the order of that list using Python list slicing ([::-1]) or the .reverse() method.

Opens a new file called reversed_boot.txt in "w" mode and dumps the reversed list into it using .writelines().

```python
with open("boot_order.txt","r") as f1, open("reversed_boot.txt","w") as f2:
    con = f1.readlines()
    rev = con[::-1]
    f2.writelines(rev)
```
