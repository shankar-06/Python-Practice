### DataFrames
DataFrames are the bread and butter of Pandas, and they are essentially just tables, like a spreadsheet or a SQL table.

#### Pandas DataFrames: Explained Simply
While a Pandas Series is a single column of data (1-Dimensional), a DataFrame is a full table with multiple columns and multiple rows (2-Dimensional).

Here are the key concepts taught in the video:  

##### Creating a DataFrame: 
The easiest way to start is by building a Python dictionary. The keys of the dictionary become the column headers, and the values (which are lists) become the data in the rows below. You then pass that dictionary into the pd.DataFrame() constructor.  
##### Custom Indexing: 
Just like in a Series, rows are automatically labeled starting from 0. You can pass a custom list of row labels using the index= argument so your rows have meaningful names.
##### Locating Rows (.loc / .iloc): 
If you want to grab the data for one specific row, you use .loc[] (location by your custom label) or .iloc[] (integer location by its numerical index, like 0, 1, 2).
##### Adding a Column: 
You can create a new column just like you add a key to a dictionary. For example, df['job'] = ['cook', 'cashier', 'manager'] assigns a new column to the right side of your table.
##### Adding a Row (pd.concat): 
To add new rows, you create a second, smaller DataFrame representing the new data, and then you use pd.concat() to glue the two DataFrames together.
