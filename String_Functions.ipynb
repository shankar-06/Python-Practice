### String Functions
##### 1. Case Changing & Formatting
These methods alter how the text looks without changing the actual letters.

.capitalize(): Capitalizes only the very first letter of the string (e.g., 'Hello this is python').

.upper(): Converts every letter to UPPERCASE.

.lower(): Converts every letter to lowercase.

.casefold(): Similar to .lower(), but stronger. It is designed to aggressively convert all characters to lowercase, which is especially useful for matching text in different languages.

.title(): Capitalizes the first letter of every word in the string.

.swapcase(): Flips the case of every letter (uppercase becomes lowercase, and vice versa).

.center(width): Centers the string within a given width (though your note uses .center(1) and .center(-1), which won't change much if the string is already longer than 1 character).

##### 2. Searching & Counting
These methods help you find specific information within a string.

.count("substring"): Returns how many times a specific character or phrase appears. In your notes, word.count("s") returned 0 because "s" was lowercase in your search but uppercase in the actual string ("HELLo tHIS iS pYTHON").

.find("substring"): Returns the index (position) of the first time it finds the requested substring.

.endswith("suffix"): Checks if the string ends with a specific character or phrase (returns True or False).

##### 3. Boolean Checks (Validation)
These methods return True or False to validate the contents of a string.

.isdigit(): Checks if all characters in the string are numbers (e.g., "12".isdigit() is True).

.isalpha(): Checks if all characters are letters (spaces will cause this to return False).

.isalnum(): Checks if characters are alphanumeric (letters AND numbers).

.isspace(): Checks if the string consists only of whitespace.

.isprintable(): Checks if all characters in the string are printable (useful for catching hidden control characters).

##### 4. Modifying & Splitting
These are crucial for data cleaning and extraction.

.replace("old", "new"): Swaps a specific character/phrase for another. Your notes show replacing "o" with "0", turning " python " into " pyth0n ".

.strip(): Removes leading and trailing whitespace from the ends of the string.

.removeprefix("prefix"): Specifically targets and removes a string from the very beginning.

.removesuffix("suffix"): Specifically targets and removes a string from the very end.

.split(): Breaks a string into a list of words, usually splitting wherever there is a space.

.splitlines(): Breaks a multi-line string (created with """) into a list, where each line becomes a separate item in the list.

##### 5. Encoding
.encode(): Converts the standard string into a byte string (e.g., b' pyth0n '). This is often used when preparing text data to be sent over a network or saved to specific file types.
