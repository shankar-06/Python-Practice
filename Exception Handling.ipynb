### Error Handling.

Instead of letting your entire program crash and show an ugly red traceback when something goes wrong (like trying to divide by zero), error handling allows you to catch the issue, manage it gracefully, and keep the program running.

1. try:
* This is where you place the code that you suspect might cause an error.  In your div(a,b) function, the try block contains return a/b. Python will attempt to run this first.

2. except:
* If the code inside the try block fails, Python immediately stops what it's doing and jumps to the except block to see if you have instructions for that specific error.
* You can stack multiple except blocks to handle different types of errors.
* When you ran div(5, 0), Python encountered a math error, skipped the rest of the code, and jumped to except ZeroDivisionError:, printing "ZeroDivisionError".
* When you ran div(5, "s"), Python encountered a type mismatch (dividing an integer by a string), jumped to except TypeError:, and printed "TypeError".

3. else:
* The else block runs only if the try block succeeds completely without triggering any errors.
* In your notes, this block contains print("Process Done").
* An important quirk to notice: When you ran div(10, 2) successfully, "Process Done" did not print. This is because you used the return keyword inside the try block. return instantly exits the function, so it skipped the else block entirely! If you had stored the calculation in a variable (like c = a/b) instead of returning it, the else block would have executed perfectly.

4. finally:
* The finally block runs no matter what—whether the try block succeeded, an error was caught, or the function was exited early.
* It is typically used for cleanup tasks, like closing a database connection or saving a file.
* Notice how "Error Check Was Completed" successfully printed in every single one of your test executions: div(5,"s"), div(5,0), and div(10,2).  

1. Write a basic function called add_items(a, b) that contains a try block attempting to print a + b. Add an except TypeError block that prints "You can only add identical data types!" to handle instances where a user passes an integer and a string.

```python
def add_items(a,b):
    try:
        print(a+b)
    except TypeError:
        print("You can only add identical data types!")
add_items(10,"e")
```

2. Modify your add_items function to include a finally block that prints "Addition attempt finished." regardless of whether the addition worked or failed.

```python
def add_items(a,b):
    try:
        print(a+b)
    except TypeError:
        print("You can only add identical data types!")
    finally:
        print("Addition attempt finished.")
add_items(10,10)
```

3. Write a function called divide_safe(a, b). Inside the try block, calculate result = a / b (do not use return). Add an except ZeroDivisionError block. Then, add an else block that prints "Division successful!".

```python
def divide_safe(a,b):
    try:
        result = a/b
        print(result)
    except ZeroDivisionError:
        print("Division Failed!")
    else:
        print("Division successful!")
divide_safe(10,2)
```

4. Error handling is great for lists! Write a try block that attempts to print the 10th item of a list my_list = [1, 2, 3]. Because this index doesn't exist, write the specific except IndexError block to catch it and print "That item does not exist in the list."

```python
try:
    my_list = [1,2,3]
    my_list[10]
except IndexError:
    print("That item does not exist in the list.")
```

5. Based on the quirk in your notes, what happens if you place a return statement inside the try block, the except block, AND the finally block? Which return value actually gets sent back to the user when the function finishes? (Hint: test this one out in your Jupyter Notebook to see the priority order!)

```python
def unsafe_div(a,b):
    try:
        result = a/b
        return(result)
    except ZeroDivisionError:
        print("Division Failed!")
    else:
        print("Division successful!")
    finally:
        print("Attempt Done!")
unsafe_div(10,2)
```
