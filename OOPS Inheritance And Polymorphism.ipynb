### Single Inheritance
#### Question 1
Write a base class BankAccount with instance attributes account_no and balance. Create a child class called SavingsAccount that inherits from BankAccount. Add a new attribute interest_rate to the child class and a method add_interest() that calculates the interest and adds it to the balance. Write the driver code to test the child class.

```python
class BankAccount():
    def __init__(self, account_no, balance):
        self.account_no = account_no
        self.balance = balance
class SavingsAccount(BankAccount):
    def __init__(self, account_no, balance, interest_rate):
        super().__init__(account_no, balance)
        self.interest_rate = interest_rate
    def add_interest(self):
        self.balance += self.balance * (self.interest_rate / 100)
        return self.balance
```

```python
q1 = SavingsAccount("SBI47339",12000,2)
print("Updated Balance: ",q1.add_interest())
```

### Multilevel Inheritance
#### Question 4
Create a base class SecuritySystem with an attribute system_id. Create a child class Firewall that inherits from SecuritySystem and adds a method block_ip(ip_address). Create a grandchild class IntrusionDetectionSystem that inherits from Firewall and adds a method detect_anomaly(). Instantiate the grandchild class and call all available methods.

```python
class SecuritySystem():
    def __init__(self, system_id):
        self.system_id = system_id
class Firewall(SecuritySystem):
    def __init__(self, system_id, ip_address):
        super().__init__(system_id)
        self.ip_address = ip_address
    def block_ip(self):
        print(f"The IP Address {self.ip_address} have been blocked")
class IntrusionDetectionSystem(Firewall):
    def __init__(self, system_id, ip_address, warning):
        super().__init__(system_id,ip_address)
        self.warning = warning
    def detect_anomaly(self):
        print(f"There has been Warning detected at {self.warning}")
```

```python
q4 = IntrusionDetectionSystem("VY2829", "102:8322:9404", "Traffic Log")
q4.block_ip()
q4.detect_anomaly()
```
