### Nested Loop Assignment

1. Hollow Square Write a Python program to print a hollow square of * with side length n.

```python
row = 6
for i in range(1, row+1):
    for j in range(1, row+1):
        if i == 1 or i == row or j == 1 or j == row:
            print("*",end = " ")
        else:
            print(" ",end=" ")
    print()
```

2. Hollow Rectangle Print a hollow rectangle of * with given rows and columns.

```python
row = 4
col = 8
for i in range(1, row+1):
    for j in range(1, col+1):
        if i == 1 or i == row or j == 1 or j == col:
            print("*",end=" ")
        else:
            print(" ",end=" ")
    print()
```

3. Hollow Right-Angled Triangle Print a hollow right-angled triangle of * with height n.

```python
tri = 6
for i in range(1, tri + 1):
    for space in range(tri - i):
        print(" ",end = " ")
    for j in range(1, i+1):
        if j == 1 or j == i or i == tri:
            print("*",end = " ")
        else:
            print(" ",end=" ")
    print()
```

```python
num = 6
for i in range(num,0,-1):
    for space in range(num - i):
        print(" ",end = " ")
    for j in range(1, i +1):
        if j == 1 or j == i or i == num:
            print("*",end=" ")
        else:
            print(" ",end=" ")
    print()
```

9. Hollow Cross (Plus Sign) Print a hollow cross + pattern inside an n × n grid.

```python
n = 9
mid = (n + 1) // 2
for i in range(1,n+1):
    for j in range(1,n+1):
        if j == mid or i == mid:
            print("*",end=" ")
        else:
            print(" ",end=" ")
    print()
```

10. Hollow X (Diagonal Halo) Print a hollow X pattern (stars only on diagonals) inside an n × n grid.

```python
n = 9
for i in range(1,n+1):
    for j in range(1,n+1):
        if i == j or i+j == n+1:
            print("*",end=" ")
        else:
            print(" ",end=" ")
    print()
```
