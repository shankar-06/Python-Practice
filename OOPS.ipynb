Using the Employees class from your notes, write a new method called apply_bonus(self, percentage). This method should calculate the bonus based on the employee's current salary, add it to their total salary, and update the self.salary attribute.

```python
class Employees:
    company = "Umbrella Corporation"
    def __init__(self, emp_id, name, dept, salary):
        self.emp_id = emp_id
        self.name = name
        self.dept = dept
        self.salary = salary
    def change_dept(self,transferred_dept):
        self.dept = transferred_dept
    def add_salary(self,new_salary):
        self.salary += new_salary
    def describe(self):
        return f" The Employee {self.name} From {self.dept} Earns {self.salary} at {self.company}"
    def apply_bonus(self, percentage):
        self.salary += self.salary * (percentage / 100)
e1 = Employees(103, "Theo", "DE", 45000)
e1.apply_bonus(18)
print(e1.describe())
```

Question 3
Create a BankAccount class. It should have a class attribute called bank_name set to "Standard Chartered". Its __init__ method should initialize account_holder and balance. Write a withdraw(self, amount) method that deducts the amount from the balance, but only if the balance is greater than or equal to the withdrawal amount.

```python
class BankAccount:
    bank_name = "Standard Chartered"
    def __init__(self, account_holder, balance):
        self.account_holder = account_holder
        self.balance = balance
    def withdraw(self,amount):
        if self.balance >= amount:
            self.balance -= amount
        else:
            print("Insufficient balance for withdrawal.")
    def display(self):
        return f" Bank Name : {self.bank_name}, Account Holder : {self.account_holder}, Balance : {self.balance}"
```

```python
q3 = BankAccount(433534352,1800)
q3.display()
```

```python
q3.withdraw(800)
q3.display()
```

```python
class ShoppingCart:
    def __init__(self):
        self.items = []
    def add_item(self, item_name):
        self.items.append(item_name)
    def count_items(self):
        return len(self.items)
```

Question 6: Student Grading System
Write a class named Student.

The __init__ method should initialize student_id, name, and an empty list named marks.

Write a method add_mark(self, mark) that appends an integer mark to the marks list.

Write a method calculate_average(self) that computes and returns the average score of all marks in the list (if the list is empty, return 0).

Write a method get_status(self) that returns "Passed" if the average is 50 or higher, and "Failed" otherwise.

```python
class Student:
    def __init__(self, student_id, name):
        self.student_id = student_id
        self.name = name
        self.marks = []
    def add_mark(self, mark):
        self.marks.append(mark)
    def calculate_avg(self):
        if not self.marks:
            return 0
        return sum(self.marks) / len(self.marks)
    def get_status(self):
        if self.calculate_avg() >= 50:
            return "Passed"
        return "Failed"
```

```python
s1 = Student(101, "Alex")
s1.add_mark(60)
s1.add_mark(70)
s1.add_mark(45)
print("Average:", s1.calculate_avg())
print("Status:", s1.get_status())
```

Question 7: Book Inventory Tracker
Write a class named Book.

Define a class attribute called library_name set to "City Central Library".

The __init__ method should accept and initialize title, author, and stock_count.

Write a method borrow_book(self) that decreases stock_count by 1, but only if the stock is greater than 0. If it is 0, print "Out of stock".

Write a method return_book(self) that increases the stock_count by 1 and prints a thank you message.

```python
class Book:
    library_name = "City Central Library"
    def __init__(self, title, author, stock_count):
        self.title = title
        self.author = author
        self.stock_count = stock_count
    def borrow_book(self):
        if self.stock_count > 0:
            self.stock_count -= 1
        else:
            print("Out Of Stock")
    def return_book(self):
        self.stock_count += 1
        print("Thank You")
```

Question 8: E-commerce Product Discount
Write a class named Product.

The __init__ method should initialize product_id, product_name, and price.

Write a method apply_discount(self, discount_percent) that reduces the price by the given percentage and permanently updates the self.price attribute.

Write a method display_info(self) that returns a formatted string containing the product's name and its current price.

Bonus: Create an object for a product that costs 2000, apply a 15% discount, and print its info.

```python
class Product:
    def __init__(self,product_id, product_name, price):
        self.product_id = product_id
        self.product_name = product_name
        self.price = price
    def apply_discount(self, discount_percent):
        self.price -= self.price * (discount_percent / 100)
    def display_info(self):
        return f"Product Name: {self.product_name} | Final Price: {self.price}"
```

```python
p1 = Product(27211,"Kinder Joy Set",2000)
p1.display_info()
```

```python
p1.apply_discount(15)
p1.display_info()
```
