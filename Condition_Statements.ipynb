*   If it is 50 or more, print "Notice: 50% threshold reached."

```python
daily_limit = int(input("Enter The Daily Limit: "))
current_spent = int(input("Enter The Current Spending: "))
percentage_spent = (current_spent / daily_limit) * 100
if percentage_spent >= 100:
    print("Alert: 100% threshold reached")
elif percentage_spent >= 80:
        print("Warning: 80% threshold reached")
elif percentage_spent >= 50:
            print("Notice: 80% threshold reached")
else:
    print("Spending within safe limits")
```
