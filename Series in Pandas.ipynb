##### Question 1: Lists & Custom Indexing
Create a Python list containing the software stability percentages of four androids (e.g., 98, 85, 45, 92). Convert this list into a Pandas Series. Instead of the default numeric index, pass in a custom index using character names from Detroit: Become Human (e.g., 'Connor', 'Markus', 'Kara', 'Chloe'). Print the final Series.

```python
import pandas as pd
software_stability_percentages = [98,85,45,92]
series1 = pd.Series(software_stability_percentages, index = ("Connor","Markus","Kara","Chloe"))
print(series1)
```

##### Question 2: Dictionaries & Updating with .loc
Create a Python dictionary representing the number of frames won by different players in a local snooker tournament. Use the player names as keys and the frames won as values. Convert this dictionary into a Pandas Series. One of the players just won two more back-to-back frames. Using the .loc property, update that specific player's score by adding 2 to their total. Print the updated Series.

```python
import pandas as pd
players = {"Theo":12,"Dinesh":15,"Murali":10}
series2 = pd.Series(players)
series2.loc["Murali"] += 2 
print(series2)
```

##### Question 3: Locating Values (.iloc vs .loc)
Create a Pandas Series containing the number of flagged network anomalies detected by an Intrusion Detection System across 4 different firewall ports (e.g., 'Port_21', 'Port_22', 'Port_80', 'Port_443').

First, write the code to retrieve the anomaly count of the very first item in the Series using its integer location.

Next, write the code to retrieve the anomaly count of 'Port_80' using its label location.

```python
import pandas as pd
IDS = {"Port_21":2,"Port_22":6,"Port_80":43,"Port_443":12}
series3 = pd.Series(IDS)
print(series3.iloc[0])
print(series3.loc["Port_80"])
```

##### Question 4: Filtering by Value
Create a Pandas Series representing daily transaction volumes for a branch of ICICI Bank over 5 days. Use a list of daily transaction counts (e.g., 1500, 2200, 1800, 3100, 1950) and assign day labels ('Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday') as the index. Write a condition using the subscript operator [] to filter and print only the days where the transaction volume was greater than or equal to 2000.

```python
import pandas as pd
TC = [1500,2200,1800,3100,1950]
series4 = pd.Series(TC,index = ("Monday","Tuesday","Wednesday","Thursday","Friday"))
print(series4[series4 >= 2000])
```
