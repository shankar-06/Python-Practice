### The Python Function (def) Blueprint
Whenever you build a function, you are no longer writing a one-time script; you are building a reusable factory machine. Here is exactly how it works.

#### 1. The Syntax (The Anatomy of the Machine)

The def keyword: This tells Python, "I am building a machine."

The Name: Written in lowercase with underscores (e.g., calculate_tax).

The Parameters ( ): The input slots. This is the raw data (like numbers or lists) you feed into the machine.

The return statement: The most important part. This is the output chute. It physically hands the calculated data back to the system. (Note: print() just flashes text on a monitor for humans; return gives the actual data back to the computer).

#### 2. The Order of Execution (The Timeline)
Python does not read functions from top to bottom like normal code.

Phase 1 (Memorization): When Python sees the def block, it memorizes the code inside it, but it does not run it. The machine is built, but turned OFF.

Phase 2 (The Call): Python keeps reading your normal code until it sees you "call" the function's name (e.g., my_answer = calculate_tax(100)).

Phase 3 (Execution): The machine turns ON. Python pauses your main code, jumps inside the function, plugs the 100 into the parameter, and runs the math.

Phase 4 (The Return): The function hits return, spits out the final answer, shuts back off, and Python goes right back to where it paused in Phase 2.

#### 3. The 3 Core Function Patterns for Data Engineers

The Math Transformer: Takes a raw number, applies a formula, returns a clean number (e.g., Celsius to Fahrenheit).

The Validator: Takes a string or number, checks it against rules, and returns a Boolean (True or False). Heavily used in data quality checks.

The Loop Wrapper: Takes a raw list, creates an empty clean_list = [] inside the function, loops through the raw data, appends the good stuff, and returns the new list.

Question 1: The Factorial Transformer
Write a function named get_factorial that takes one parameter (num). Inside the function, use a for loop to calculate the factorial of that number. Return the final total.

Hint: Create a total = 1 variable inside the function. Loop from 1 up to num + 1, multiplying the total by i.

```python
def get_factorial(num):
    total =1
    for i in range(1,num+1):
        total *= i
    return total
user = int(input("Enter The Factorial Number: "))
print(get_factorial(user))
```

Question 2: The Vowel Counter
Write a function named count_vowels that takes one parameter (text). The function should count exactly how many vowels (a, e, i, o, u) are in the string, and return that count.

Hint: Create a count = 0 variable. Use for letter in text: and an if letter in "aeiouAEIOU": statement to add to your counter.

```python
def count_vowels(text):
    count = 0
    for i in text:
        if i == "A" or i == "a" or i == "E" or i == "e" or i == "I" or i == "i" or i == "O" or i == "o" or i == "U" or i == "u": 
            count +=1
    return count
user = input("Enter The Text: ")
print(count_vowels(user))
```

```python
# Alternative for Question 2: 

def count_vowels(text):
    count = 0
    for i in text:
        if i in "aeiouAEIOU":
            count +=1
    return count
user = input("Enter The Text: ")
print(count_vowels(user))

# This tells Python: "Check if the letter i matches any of the letters inside this string." 
# It does the exact same thing as your long if statement, but saves you a ton of typing!
```

Question 3: The Palindrome Validator
Write a function named is_palindrome that takes one parameter (word). A palindrome is a word that is spelled the same forwards and backwards (like "radar" or "level"). If it is a palindrome, return True. If not, return False.

Hint: You don't need a loop! You can use Python's string slicing trick word[::-1] to reverse a word. Just check if word == word[::-1]:.

```python
def is_palindrome(text):
    text1 = text.lower()
    if text1 == text1[::-1]:
        return True
    else:
        return False
user = input("Enter The Palindrome: ")
check = is_palindrome(user)
if check == True:
    print("It is Palindrome")
else:
    print("The String is not Palindrome")
```

Question 4: The Fibonacci Generator
Write a function named generate_fibonacci that takes one parameter (terms). Instead of just printing the numbers, the function should generate the sequence and return it as a list.

Hint: Start with fib_list = [], a = 0, and b = 1. Write a for loop that runs terms times. Inside the loop, fib_list.append(a), then do the shift trick: a, b = b, a + b. Return the list at the end.

```python
def generate_fibonacci(terms):
    fib_list = []
    a = 0
    b = 1
    for i in range(terms):
        fib_list.append(a)
        a,b = b, a + b
    return fib_list    
generate_fibonacci(12)
```
