### Database Connection
#### Step 1: Installing the Required Package
* The notebook begins by running pip install mysql-connector-python.
* This command downloads and installs the official driver that allows Python to communicate with MySQL.
* The output confirms that version 26.7.0 was successfully installed.

#### Step 2: Importing the Module
* The import mysql.connector statement loads the installed library into your Python environment so its functions can be used.
* A simple print("Done") command is executed to visually confirm that the import was successful without any errors.

#### Step 3: Establishing the Database Connection
* The code uses mysql.connector.connect() to create a connection object named conn.
* It connects to the local machine using the IP address 127.0.0.1.
* It logs in using the default database administrator credentials: username "root" and an password "".
* It explicitly connects to a database named Student_Course_Management_System.
* Upon successfully connecting, it prints "Database Connection Done".

#### Step 4: Initializing the Cursor
* The command cursor = conn.cursor() initializes a cursor object.
* In database programming, a cursor is a control structure that acts as a middleman. It is responsible for sending your SQL statements to the database and temporarily holding the results that the database sends back.

#### Step 5: Executing the SQL Query
* A standard SQL query string, "select * from Students", is assigned to the variable select.
* The cursor.execute(select) method is called to send this query to the MySQL database. The database processes the request and prepares the data, but does not send it directly to the Python screen yet.

#### Step 6: Fetching and Displaying the Data
* The code uses a for loop combined with cursor.fetchall().
* The fetchall() method grabs every single row of data that the database prepared in Step 5 and returns it as a list of Python tuples.
* The loop iterates through this list, printing each row one by one.
* The output displays 10 rows of student records. Each row contains the student's ID, full name, date of birth (formatted as a datetime.date object), gender, and email address.
