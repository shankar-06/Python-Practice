### Loop Statement

1. Sum of Numbers
Write a program to calculate the sum of all numbers from 1 to 100 using a for loop.

```python
total = 0
for i in range(1,101):
    total += i
print(total)
```

2. Table Generator
Take a number as input from the user and print its multiplication table (1 to 10) using a while loop.

```python
user = int(input("Enter The Number: "))
multiples = 0
answer = 0
while multiples < 10:
    multiples += 1
    answer = user * multiples
    print(f"{user} * {multiples} = {answer}")
```

3. Even/Odd Counter
Given a list of numbers, count how many are even and how many are odd using a for loop.

```python
num = 25
Even = []
Odd = []
for i in range(1,num+1):
    if i%2 == 0:
        Even.append(i)
    else:
        Odd.append(i)
print(f"Even Numbers: {Even}")
print(f"Odd Numbers: {Odd}")
```

4. Reverse a Number
Write a program that reverses a given integer using a while loop (e.g., 1234 → 4321).

```python
num = int(input("Enter The Number: "))
rev_num = 0
while num > 0:
    last_num = num % 10
    rev_num = rev_num * 10 + last_num
    num //= 10
print(f"The Reverse Number is {rev_num}")
```

5. Prime Number Check
Write a program to check whether a given number is prime using a for loop.

```python
num = int(input("Enter The Number: "))
sqrt_num = int(num**0.5)
count = 0
for i in range(1,sqrt_num+1):
        if num% i == 0:
            count += 1
if count == 1:
    print(f"The Number {num} You Gave Is An Prime Number")
else:
    print(f"The Number {num} You Gave Is Not Prime Number")
```

6. Star Pattern
Print the following pattern using nested loops:

```python
row = 5
for i in range(1,row+1):
    for j in range(1, i+1):
        print("*",end="")
    print()
```

7. Factorial Using while
Calculate the factorial of a number entered by the user using a while loop.

```python
num = int(input("Enter The Factorial Number: "))
fact = 1
while num > 0:
    fact *= num
    num -=1
print(fact)
```

8. Skip Multiples with continue
Print numbers from 1 to 20, but skip all multiples of 3 using continue.

```python
for i in range(1,21):
    if i%3 == 0:
        continue
    print(i)
```

9. Stop at First Match with break
Given a list of numbers, write a program that stops looping and prints a message as soon as it finds the first number greater than 50.

```python
temp_list = [12,30,7,45,2,0,50,83,15]
for i in temp_list:
    if i >= 50:
        print("The Number is Greater than 50")
        break
    print(i)
```

10. Fibonacci Series
Write a program to print the first n terms of the Fibonacci series using a for or while loop.

```python
num = int(input("Enter The Number: "))
fib_list = []
a = 0
b = 1
for i in range(num+1):
    fib_list.append(a)
    a,b = b, a+b
print(fib_list)
```
