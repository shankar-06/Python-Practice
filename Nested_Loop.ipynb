### The 4-Step Pattern Cracking Blueprint

Whenever you see a pattern in an interview, do not start typing. Ask yourself these four questions in this exact order:

1. The Outer Loop (The Height)Question: How many rows (lines) are going top-to-bottom?Code Translation: This is your for i in range(...). If there are 5 rows, your outer loop is always range(1, 6).
   
2. The Inner Loop (The Width Rule)Question: Look at Row 1, then Row 2, then Row 3. What is happening to the width? Is it growing, shrinking, or staying the same?Code Translation: This is your for j in range(...). You must link j to i.If it grows (1 star, then 2 stars, then 3): The inner loop is range(1, i + 1).  If it shrinks (3 stars, then 2 stars, then 1): The outer loop counts backward range(row, 0, -1) and the inner loop stays range(1, i + 1).  If it forms a box (3 stars on every single row): The inner loop is completely disconnected from i, like range(1, row + 1).


3. The Payload (What are we printing?)Question: Are we printing a symbol, a row tracker, or a column tracker?Code Translation:If it is a shape: print("*", end="").  If the number stays the same across the row (22, 333): print(i, end="").  If the number counts up across the row (12, 123): print(j, end="").

4. The Left Edge (Do we need spaces?)Question: Look at the left side of the shape. Is it a flat, straight wall, or is it slanted?Code Translation: If it is a straight wall, you only need one inner loop. If it is slanted to the right, you must write a Space Loop before your character loop to push the characters over (for space in range(row - i): print(" ", end="")).


Here are 6 nested loop pattern challenges. I have included small hints based strictly on the 4-Step Visual Decoding Blueprint to help guide your logic without giving away the exact code.

Assume the user enters row = 5 for all of these.

Challenge 1: The Square Grid
*****
*****
*****
*****
*****
Hint: Look at Step 2 (The Width Rule). The width does not grow, and it does not shrink. Every single row has exactly 5 stars. How do you make the inner loop run 5 times, no matter what row the outer loop is on?

```python
row = 5
for i in range(1,row+1):
    for j in range(1,row+1):
        print("*",end="")
    print()
```

Challenge 2: The Row Tracker
1
22
333
4444
55555
Hint: Look at Step 3 (The Payload). The width is growing, so the inner loop is standard. But notice the numbers. On Row 3, it prints 333. The number does not change as it goes across the column. Which loop variable (i or j) stays the same across the entire row?

```python
row = 5
for i in range(1,row+1):
    for j in range(1,i+1):
        print(i,end="")
    print()
```

Challenge 3: The Inverted Star Triangle
*****
****
***
**
*
Hint: Look at Step 1 (The Height) and Step 2 (The Width). The shape is shrinking. The easiest way to shrink a shape is to leave the inner loop exactly the same, but force the outer loop (the hour hand) to count backwards using a negative step in your range().

```python
row = 5
for i in range(row,0,-1):
    for j in range(1,i+1):
        print("*",end="")
    print()
```

Challenge 4: The Inverted Column Tracker
12345
1234
123
12
1
Hint: This combines the logic of Challenge 2 and Challenge 3. The shape is shrinking (so the outer loop must go backwards), but the numbers are changing as they go across the row (1, then 2, then 3). Which variable tracks the column moving across?

```python
row = 5
for i in range(row,0,-1):
    for j in range(1,i+1):
        print(j,end="")
    print()
```

Challenge 5: The Centered Pyramid
        * 
      * * * 
    * * * * * 
  * * * * * * * 
* * * * * * * * * 
Hint: Look at Step 4 (The Left Edge). It is slanted, which means you need two inner loops side-by-side. The first inner loop pushes invisible spaces (row - i), and the second inner loop prints the stars. Remember the math trick from your notes to generate an odd number of stars: (1, 2*i).

```python
row = 5
for i in range(1,row+1):
    for j in range(row-i):
        print(" ",end=" ")
    for k in range(1,2*i):
        print("*",end=" ")
    print()
```

Challenge 6: The Diamond
        * 
      * * * 
    * * * * * 
  * * * * * * * 
* * * * * * * * * 
* * * * * * * * * 
  * * * * * * * 
    * * * * * 
      * * * 
        * 
Hint: Do not overthink this! A diamond is not a brand new loop trick. It is simply Challenge 5 printed once, followed immediately by Challenge 5 printed completely upside down. You will need two completely separate outer loops stacked on top of each other.

```python
row = 5
for i in range(1,row+1):
    for k in range(row - i):
        print(" ",end=" ")
    for j in range(1,2*i):
        print("*",end=" ")
    print()
for i in range(row,0,-1):
    for k in range(row-i):
        print(" ",end=" ")
    for j in range(1,2*i):
        print("*",end=" ")
    print()
```
