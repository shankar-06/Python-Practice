Having a systematic approach to data cleaning is exactly what hiring managers look for in Data Engineering interviews. They want to see that you do not just blindly execute code, but that you investigate, plan, and standardize.

Here is your universal blueprint for tackling any raw dataset.

## The Universal Data Cleaning Game Plan
### Phase 1: Assess & Profile (The Reconnaissance)
Never modify data before understanding its shape. Your first goal is to spot anomalies, missing values, and incorrect data types.

* Check the footprint: How many rows and columns exist? Are numbers accidentally stored as strings?

* Identify missingness: Are null values random, or is an entire column completely empty?

* Spot weird placeholders: Look for "UNKNOWN", "N/A", "9999", or "ERROR" masquerading as valid data.

### Phase 2: Structural & Type Cleaning (The Foundation)
* Once you know what is broken, fix the structural shell of the DataFrame before altering individual cell values.

* Standardize headers: Convert all column names to lowercase snake_case (e.g., Price Per Unit to price_per_unit) so they are database-ready.

* Drop pure duplicates: Remove identical rows immediately to prevent skewed downstream aggregations.

* Cast strict types: Force columns into their proper formats (e.g., converting a string date into a DateTime object, or a string price into a float).

### Phase 3: Text & Value Standardization (The Polish)
* This is where you use your Regex and string manipulation skills to clean the actual data payload.

* Impute or drop nulls: Decide whether to fill missing values with a default/median value or drop the row entirely.

* Vectorized string cleaning: Strip hidden whitespace and standardize capitalization across all categorical text.

* Unify categories: Group similar variations together (e.g., merging "CreditCard", "credit card", and "CC" into a single "credit_card" category).

import pandas as pd

* A purposefully messy dataset
data = {

}
df = pd.DataFrame(data)
1. Assessing the Data (The Detective Work)
Before you change anything, you must understand what is broken.

df.info()

What it does: Gives you the "blueprint" of your DataFrame. It tells you the total number of rows, the exact column names, how many non-null (non-empty) values exist in each column, and the data types.

Why it's crucial: If your "Age" column is supposed to be numbers, but .info() says it is an object (string), you instantly know you have bad data hidden in there (like the word "ERROR").

Python
df.info() 
df['column'].value_counts()

What it does: Counts how many times each unique value appears in a specific column.

Why it's crucial: It exposes weird placeholders. If you run this on a "Status" column and see Success: 500, Failed: 20, and 999: 5, you instantly know 999 is a garbage value that needs cleaning.

Python
df[' Age'].value_counts() # Will reveal the word 'ERROR' hiding among numbers

2. Structural Cleaning (Fixing the Shell)
Now we fix the shape and structure of the table itself.

df.rename(columns={})

What it does: Changes column headers using a dictionary where the "key" is the old name and the "value" is the new name.

Why it's crucial: Notice how our dummy data has Cust_Name  (with a trailing space) and  Age (with a leading space)? Trying to code with those spaces is a nightmare.

Python
* Rename columns to standard lowercase snake_case
df = df.rename(columns={
    'Cust_Name ': 'cust_name', 
    ' Age': 'age', 
    'Join Date': 'join_date'
})
df.drop_duplicates()

What it does: Scans the entire DataFrame and completely removes any row that is an exact, 100% clone of another row.

Why it's crucial: Duplicate rows will mathematically ruin aggregations (like sums and averages).

Python
* Drops the second 'Alice' row because it is an exact clone of the first
df = df.drop_duplicates()
 
3. Value Cleaning (Fixing the Payload)
This is where we replace bad placeholders and handle the missing gaps.

df['column'].replace()

What it does: Swaps specific targeted values for new ones.

Why it's crucial: We need to turn words like "ERROR" or "UNKNOWN" into actual math-friendly null values (np.nan) so Pandas knows they are officially blank.

Python
* Replace string placeholders with actual Null values
df['age'] = df['age'].replace('ERROR', np.nan)
df['join_date'] = df['join_date'].replace('UNKNOWN', None)
df['column'].fillna()

What it does: Finds every officially blank cell (NaN / None) and fills it with a value of your choice.

Why it's crucial: If you try to do math on a column with NaN values, it will often break. You can fill gaps with a hardcoded number (like 0), or a calculated metric.

Python
* Fill missing ages with the number 0
df['age'] = df['age'].fillna(0)

4. Text & Type Cleaning (The Final Polish)
Now we fix the messy capitalization and force the columns into the correct data formats.

df['column'].str.strip() and .str.title()

What it does: .strip() deletes invisible spaces at the start and end of strings. .title() capitalizes the first letter of every word.

Why it's crucial: In a database, "eve" and "Eve " are considered two entirely different people. We must standardize them.

Python
* Method Chaining: Strip spaces, then title case the names
df['cust_name'] = df['cust_name'].str.strip().str.title()
df['column'].astype()

What it does: Forces a column to become a specific mathematical data type (like an integer int or a decimal float).

Why it's crucial: Now that we replaced the word "ERROR" and filled our nulls, we can safely convert the Age column from text strings into actual numbers.

Python
df['age'] = df['age'].astype(int)
pd.to_datetime()

What it does: Converts text strings that look like dates into actual, calculable Pandas Timestamp objects.

Why it's crucial: If dates are left as strings, you cannot sort them chronologically or calculate the difference between two dates (like "days since joined").

Python
df['join_date'] = pd.to_datetime(df['join_date'])
