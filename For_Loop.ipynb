### For Loop
In Data Engineering, you don't need overly complicated loop tricks; you just need rock-solid logic. Here is exactly how these three methods translate directly into your future daily tasks as a Data Engineer:

1. Iterating through a List: This is the most common loop you will ever write. In the real world, instead of a list of fruits ["Apple", "Banana"], you will be iterating through a list of raw data files, a list of customer records, or a list of API responses to clean them one by one.

2. Using the range() function: You will use this heavily for "batch processing." If you have 10,000 rows of data, you might use range() to process them in manageable chunks of 1,000 at a time so you don't crash your computer's memory.

3. Loop Control (break, continue): This is your data-cleaning toolkit. If your loop is processing 100 bank transactions and hits a corrupted or blank row, you will use continue to safely skip that bad row and keep processing the rest without crashing your pipeline.

Question 1: The Number Squarer (Using range and input)
Ask the user to Enter a maximum number: . Use a for loop and the range() function to print the square of every number from 1 up to (and including) the number the user entered.

```python
user1 = int(input("Enter The Count Of Numbers you want to Generate Square Values:"))
for i1 in range(1,user1+1):
    print(i1**2)
```

Question 2: The List Sorter (Using if/else and append)
Create two empty lists: divisible_by_3 = [] and divisible_by_5 = [].
Write a for loop that checks all numbers from 1 to 30.

If the number is perfectly divisible by 3, append it to the first list.

If the number is perfectly divisible by 5, append it to the second list.
(Note: If a number is divisible by both, like 15, it is fine if your if/else logic only puts it in the first list).
Print both lists after the loop finishes.

```python
divisible_by_3 = []
divisible_by_5 = []
for i2 in range(1,30+1):
    if i2 % 3 == 0:
        divisible_by_3.append(i2)
    elif i2 % 5 == 0:
        divisible_by_5.append(i2)
print(divisible_by_3)
print(divisible_by_5)
```

Question 3: The Unlucky Number (Using Loop Control)
Write a for loop that iterates through the numbers 1 to 10.
Inside the loop, check the number. If the number is exactly 7, use a loop control statement to skip printing it, but allow the loop to keep running so it still prints 8, 9, and 10.

```python
for i3 in range(1,11):
    if i3 == 7:
        continue
    print(i3)
```

### 🐉 The Final For Loop Boss Battles
Boss Battle 1: The ETL Data Pipeline
Given the list of raw transactions: raw_data = [150, -20, 0, 300, 45, -99, 10]

Create an empty list called clean_data and a counter variable called error_count starting at 0.

Write a for loop to check every number in raw_data.

If the transaction is strictly greater than 0, append it to the clean_data list.

If the transaction is 0 or less, do not append it. Instead, add 1 to your error_count.

After the loop finishes, print clean_data and print the error_count.

```python
raw_data = [150, -20, 0, 300, 45, -99, 10]
clean_data = []
error_count = 0
for j11 in raw_data:
    if j11 > 0:
        clean_data.append(j11)
    else:
        error_count += 1
print(f" Clean Data: {clean_data}")
print(f"Error Count: {error_count}")
```

Boss Battle 2: The String Decoder
Given the string: secret_message = "D#A#T#A#B#A#S#E"

Create an empty string variable: decoded = "" (Just like you start a math total at 0, you can start a text total as empty quotes!)

Write a for loop to iterate through every character in secret_message.

If the character is exactly #, use a loop control statement to skip it.

Otherwise, add the character to your decoded string (decoded += character).

After the loop finishes, print the final decoded string.

```python
secret_message = "D#A#T#A#B#A#S#E"
decoded = ""
for j12 in secret_message:
    if j12 == "#":
        continue
    else:
        decoded += j12
print(decoded)
```

Boss Battle 3: The Revenue Target (Accumulator + Break)
Given the list of daily sales: daily_revenue = [50, 100, 200, 300, 50, 400]

Set a variable total_revenue = 0 and a variable days_taken = 0.

Write a for loop to iterate through the daily_revenue list.

Inside the loop, add the day's revenue to total_revenue, and add 1 to days_taken.

Check if the total_revenue is greater than or equal to 500. If it is, completely break the loop.

After the loop finishes, print the total_revenue and print days_taken.

```python
daily_revenue = [50, 100, 200, 300, 50, 400]
total_revenue = 0
days_taken = 0
for j13 in daily_revenue:
    if total_revenue >= 500:
        break
    else:
        total_revenue += j13
        days_taken += 1
print(f"Total Revenue: {total_revenue}")
print(f"Total Days To Achieve more than 500 is {days_taken}")
```
