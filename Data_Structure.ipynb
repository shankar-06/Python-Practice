### What is a List?
1. 1D: It is a one-dimensional structure.  
2. Versatile: It can store multiple different data types at the same time, such as strings (e.g., "Apple"), integers (e.g., 4), and floats (e.g., 2.25).
3. Ordered: The items maintain a specific sequence.  Indexable & Sliceable: You can access specific items or ranges of items using their position.
4. Duplicates Allowed: You can store the same exact value multiple times (like having "Apple" appear twice).  
5. Mutable / Changeable: You are allowed to modify, add, or remove items after the list is created.

### What is Sets
1. 1D: It is a one-dimensional structure.  
2. Unordered: The items do not maintain a specific sequence.  
3. Duplicates NOT allowed: If you try to put the same item in twice, it simply ignores the second one.  
4. Multiple Data Types: It can store strings, integers, floats, etc.  
5. Indexing / Slicing NOT possible: Because it is unordered, there is no fixed "first" or "last" position to grab.  
6. Partially mutable / changeable: You can add or remove items, but you cannot change an existing item directly in place.
