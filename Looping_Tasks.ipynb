```python
# Amstrong Number
"""152
1^3 + 5^3 + 3^3 = 153
153 == 153 >>> Amstrong Number"""
```

```python
num = int(input("Enter Your Number: "))
c_num = num
digit = len(str(num))
result = 0
while num > 0:
    last_num = num % 10
    last_num **= digit     # last_num = last_num ** digit
    result += last_num
    num //= 10             # num = num//10
if result == c_num:
    print(f"The Number {c_num} is an Amstrong Number")
else:
    print(f"The Number {c_num} is not an Amstrong Number")
```
